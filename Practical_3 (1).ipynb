```python
# -------------------------------
# Import libraries
# -------------------------------
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
from mlxtend.preprocessing import TransactionEncoder
```

```python
# -------------------------------
# Helper function
# -------------------------------
def encode_transactions(dataset):
    te = TransactionEncoder()
    te_ary = te.fit(dataset).transform(dataset)
    return pd.DataFrame(te_ary, columns=te.columns_)
```

```python
# -------------------------------
# Apriori function
# -------------------------------
def run_apriori(df, min_support, min_confidence):
    print(f"\n--- Apriori with min_support={min_support}, min_confidence={min_confidence} ---")

    # Frequent itemsets
    freq_itemsets = apriori(df, min_support=min_support, use_colnames=True)
    print("\nFrequent Itemsets:")
    print(freq_itemsets)

    # Association rules
    rules = association_rules(freq_itemsets, metric="confidence", min_threshold=min_confidence)
    print("\nAssociation Rules:")
    print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])

    # -------------------------------
    # Evaluation summary
    # -------------------------------
    print("\nEvaluation Summary:")
    print("Support - mean:", rules['support'].mean(), "| max:", rules['support'].max())
    print("Confidence - mean:", rules['confidence'].mean(), "| max:", rules['confidence'].max())
    print("Lift - mean:", rules['lift'].mean(), "| max:", rules['lift'].max())

    return freq_itemsets, rules
```

```python
# -------------------------------
# DATASET 1
# -------------------------------
dataset1 = [
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter'],
    ['Milk', 'Bread'],
    ['Milk', 'Bread', 'Butter', 'Eggs'],
    ['Bread', 'Eggs']
]

df1 = encode_transactions(dataset1)
```

```python
# Run Apriori on Dataset 1
run_apriori(df1, min_support=0.5, min_confidence=0.75)
run_apriori(df1, min_support=0.6, min_confidence=0.6)
```

```python
# -------------------------------
# DATASET 2
# -------------------------------
dataset2 = [
    ['Apple', 'Banana', 'Orange'],
    ['Apple', 'Banana'],
    ['Apple', 'Banana', 'Orange'],
    ['Apple', 'Banana', 'Grapes'],
    ['Apple', 'Banana', 'Orange']
]

df2 = encode_transactions(dataset2)
```

```python
# Run Apriori on Dataset 2
run_apriori(df2, min_support=0.5, min_confidence=0.75)
run_apriori(df2, min_support=0.6, min_confidence=0.6)
```
