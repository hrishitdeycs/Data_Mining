```python
# =============================================
# Title: Supermarket Sales - Data Preprocessing
# Description: Apply standardization, normalization, transformation, aggregation, discretization, and sampling
# =============================================

# 1. Import Libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler, Binarizer
from google.colab import files
from IPython.display import display
```

```python
# 2. Upload Dataset
uploaded = files.upload()  # Upload CSV file
filename = list(uploaded.keys())[0]
```

```python
# 3. Load Dataset
df = pd.read_csv(filename, skipinitialspace=True)
print("Initial Dataset Sample:")
display(df.head(10))  # show first 10 rows
```

```python
# 4. Standardization (z-score) on 'Unit price' and 'Quantity'
scaler = StandardScaler()
df[['Unit price_scaled', 'Quantity_scaled']] = scaler.fit_transform(df[['Unit price', 'Quantity']])
print("\nAfter Standardization (first 5 rows):")
display(df[['Unit price_scaled', 'Quantity_scaled']].head())
```

```python
# 5. Normalization (min-max) on 'Total'
minmax_scaler = MinMaxScaler()
df['Total_normalized'] = minmax_scaler.fit_transform(df[['Total']])
print("\nAfter Normalization (first 5 rows):")
display(df[['Total', 'Total_normalized']].head())
```

```python
# 6. Transformation (log) on 'Total' to reduce skew
df['Total_log'] = np.log1p(df['Total'])  # log1p to handle 0 values
print("\nAfter Log Transformation (first 5 rows):")
display(df[['Total', 'Total_log']].head())
```

```python
# 7. Aggregation: Total sales per Branch
branch_sales = df.groupby('Branch')['Total'].sum().reset_index()
print("\nTotal sales per Branch:")
display(branch_sales)
```

```python
# 8. Discretization / Binarization of 'Rating'
# High rating (1) if Rating >= 7 else low rating (0)
binarizer = Binarizer(threshold=7)
df['High_Rating'] = binarizer.fit_transform(df[['Rating']])
print("\nAfter Binarization of Rating (first 5 rows):")
display(df[['Rating', 'High_Rating']].head())
```

```python
# 9. Sampling: Take 20% random sample of the dataset
df_sample = df.sample(frac=0.2, random_state=42)
print("\nRandom 20% Sample (first 5 rows):")
display(df_sample.head())
```
