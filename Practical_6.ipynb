```python
# Import libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score
from google.colab import files
import matplotlib.pyplot as plt
```

```python
# Upload CSV file
print("Please upload your dataset:")
uploaded = files.upload()
```

```python
# Load dataset
file_name = list(uploaded.keys())[0]
df = pd.read_csv(file_name)
print(f"\nDataset '{file_name}' loaded successfully!")
print("Columns:", df.columns)
```

```python
# If target column exists, remove it
if 'target' in df.columns:
    X = df.drop('target', axis=1)
else:
    X = df.copy()
```

```python
# Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Function to run DBSCAN and evaluate
def dbscan_evaluation(X, eps, min_samples):
    db = DBSCAN(eps=eps, min_samples=min_samples)
    labels = db.fit_predict(X)

    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
    if n_clusters > 1:
        sil_score = silhouette_score(X, labels)
    else:
        sil_score = np.nan  # silhouette undefined for 1 cluster

    print(f"DBSCAN with eps={eps}, min_samples={min_samples}")
    print(f"  Number of clusters: {n_clusters}")
    print(f"  Silhouette Score: {sil_score:.3f}")
    print(f"  Noise points: {np.sum(labels == -1)}\n")

    return labels
```

```python
# Example: vary eps and min_samples
eps_values = [0.5, 1.0, 1.5]
min_samples_values = [3, 5, 7]

for eps in eps_values:
    for min_samples in min_samples_values:
        dbscan_evaluation(X_scaled, eps, min_samples)
```
