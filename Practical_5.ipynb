```python
# Import libraries
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans

# Load dataset
data = load_iris()
X = data.data
```

```python
# Parameters to vary
cluster_options = [2, 3, 4]  # number of clusters
max_iter = 10  # iterations per run

plt.figure(figsize=(8,5))
```

```python
# Run K-Means for each cluster option
for n_clusters in cluster_options:
    inertia_list = []
    km = KMeans(n_clusters=n_clusters, init='k-means++', max_iter=1, n_init=1, random_state=42)

    for i in range(max_iter):
        km.max_iter = 1
        km.fit(X)
        inertia_list.append(km.inertia_)
        km.init = km.cluster_centers_  # continue from previous centers

    plt.plot(range(1, max_iter+1), inertia_list, marker='o', label=f"{n_clusters} clusters")
# Plot formatting
plt.xlabel("Iteration")
plt.ylabel("MSE (Inertia)")
plt.title("K-Means Convergence for Different Cluster Numbers")
plt.legend()
plt.show()
```
