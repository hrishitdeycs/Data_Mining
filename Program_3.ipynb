```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
dataset = [
    ['Milk', 'Onion', 'Nutmeg', 'Kidney Beans', 'Eggs', 'Yogurt'],
    ['Dill', 'Onion', 'Nutmeg', 'Kidney Beans', 'Eggs', 'Yogurt'],
    ['Milk', 'Apple', 'Kidney Beans', 'Eggs'],
    ['Milk', 'Unicorn', 'Corn', 'Kidney Beans', 'Yogurt'],
    ['Corn', 'Onion', 'Onion', 'Kidney Beans', 'Ice cream', 'Eggs']
]

te = TransactionEncoder()
te_ary = te.fit(dataset).transform(dataset)
df = pd.DataFrame(te_ary, columns=te.columns_)
```

```python
def run_apriori(min_support, min_confidence):
    print(f"\n--- Results for min_support={min_support}, min_confidence={min_confidence} ---")

    frequent_itemsets = apriori(df, min_support=min_support, use_colnames=True)
    print("\nFrequent Itemsets:\n", frequent_itemsets)

    rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=min_confidence)
    print("\nAssociation Rules:\n", rules[['antecedents','consequents','support','confidence','lift']])

    print("\nEvaluation Summary:")
    print("Support - mean:", rules['support'].mean(), "| max:", rules['support'].max())
    print("Confidence - mean:", rules['confidence'].mean(), "| max:", rules['confidence'].max())
    print("Lift - mean:", rules['lift'].mean(), "| max:", rules['lift'].max())
```

```python
# Part a
run_apriori(0.5, 0.75)
```

```python
# Part b
run_apriori(0.6, 0.6)
```
